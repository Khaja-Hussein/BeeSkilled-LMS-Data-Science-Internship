## Python for Data Analysis

## Question 1 - Create a Pandas DataFrame Manually

```python
import pandas as pd

data = {
    "Name": ["Rahul", "Priya", "Arjun"],
    "Marks": [85, 92, 78]
}

df = pd.DataFrame(data)
df
```

## Question 2 - Handle Missing Values with fillna()

```python
data = {
    "Name": ["Rahul", "Priya", "Arjun"],
    "Marks": [85, None, 78]
}

df = pd.DataFrame(data)

df = df.fillna(0)
df
```

## Question 3 - Sort DataFrame by a Column

```python
data = {
    "Name": ["Rahul", "Priya", "Arjun"],
    "Marks": [85, 92, 78]
}

df = pd.DataFrame(data)

df.sort_values("Marks")
```

## Question 4 - Find Mean, Median and Mode

```python
data = {
    "Math": [85, 92, 78, 85, 90],
    "Science": [88, 95, 80, 88, 92]
}

df = pd.DataFrame(data)

print("Mean:")
print(df.mean())

print("\nMedian:")
print(df.median())

print("\nMode:")
print(df.mode().iloc[0])
```
