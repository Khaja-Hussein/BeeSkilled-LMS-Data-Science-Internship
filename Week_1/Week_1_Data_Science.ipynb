## 1. NumPy

### What is NumPy?

NumPy stands for Numerical Python.

It is a Python library used for numerical computing and working with arrays.

### Why do we use NumPy?

NumPy helps us work with numerical data and perform mathematical operations efficiently.

```python
import numpy as np

numbers = np.array([10, 20, 30, 40, 50])

print(numbers)
```

```python
print(numbers + 10)
print(numbers * 2)
```

### NumPy Array Operations

NumPy allows us to perform mathematical operations directly on arrays.

```python
print(numbers - 5)
print(numbers / 10)
```

### NumPy Array Properties

NumPy arrays have useful properties that tell us about the data.

- `ndim` → number of dimensions
- `shape` → size of each dimension
- `size` → total number of elements

```python
print("Dimensions:", numbers.ndim)
print("Shape:", numbers.shape)
print("Size:", numbers.size)
```

### Accessing Array Elements

We can access individual elements of a NumPy array using their index.

Python indexing starts from `0`.

```python
print("First element:", numbers[0])
print("Third element:", numbers[2])
```

### NumPy Array Slicing

Slicing allows us to select a range of elements from an array.

The syntax is:

array[start:stop]

The `stop` index is not included.

```python
print(numbers[1:4])
print(numbers[:3])
```

## 2. Pandas

### What is Pandas?

Pandas is a Python library used for data analysis and data manipulation.

It provides useful data structures such as Series and DataFrame.

```python
import pandas as pd

data = {
    "Name": ["Ali", "Ravi", "John", "Sara"],
    "Marks": [85, 78, 92, 88]
}

df = pd.DataFrame(data)

print(df)
```

### Selecting Data from a DataFrame

We can select a single column from a Pandas DataFrame using its column name.

```python
print(df["Name"])
print(df["Marks"])
```

### Adding a New Column

We can create a new column in a DataFrame by assigning values to a new column name.

```python
df["Result"] = ["Pass", "Pass", "Pass", "Pass"]

print(df)
```

## 3. Basic Data Cleaning

### Missing Values

Missing values are empty or unavailable values in a dataset.

Before analyzing data, we need to identify and handle missing values.

```python
clean_data = {
    "Name": ["Ali", "Ravi", "John", "Sara"],
    "Marks": [85, None, 92, 88]
}

clean_df = pd.DataFrame(clean_data)

print(clean_df)
```

### Handling Missing Values with fillna()

The `fillna()` function is used to replace missing values with a specified value.

For example, we can replace a missing mark with `0`.

```python
clean_df["Marks"] = clean_df["Marks"].fillna(0)

print(clean_df)
```

### Detecting Missing Values

We can use `isnull()` to identify missing values in a DataFrame.

```python
print(clean_df.isnull())
```
